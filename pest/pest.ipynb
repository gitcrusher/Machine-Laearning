```python


def fetch_weather_nasa(lat, lon, year, month, day, base_temp=10):
    """
    Fetches daily weather data from NASA POWER API and adds
    - VPD (Vapour Pressure Deficit)
    - GDD (Growing Degree Days)
    - Season (calendar-based)
    """
    date_str = f"{year}{month:02d}{day:02d}"
    url = (
        f"https://power.larc.nasa.gov/api/temporal/daily/point?"
        f"start={date_str}&end={date_str}"
        f"&latitude={lat}&longitude={lon}"
        "&parameters=T2M_MIN,T2M_MAX,T2M,PRECTOTCORR,WS10M,WD10M,ALLSKY_SFC_SW_DWN,RH2M"
        "&community=AG"
        "&format=JSON"
    )

    try:
        r = requests.get(url, timeout=10)
        r.raise_for_status()
        data = r.json()
    except Exception:
        return None

    properties = data.get("properties", {}).get("parameter", {})
    if not properties:
        return None  # No data for this date/location

    # ----------------- Step 1: Extract core weather variables -----------------
    tmin = list(properties.get("T2M_MIN", {}).values())[0] if properties.get("T2M_MIN") else np.nan
    tmax = list(properties.get("T2M_MAX", {}).values())[0] if properties.get("T2M_MAX") else np.nan
    tmean = list(properties.get("T2M", {}).values())[0] if properties.get("T2M") else np.nan
    rh_mean = list(properties.get("RH2M", {}).values())[0] if properties.get("RH2M") else np.nan

    result = {
        "tmin": tmin,
        "tmax": tmax,
        "tmean": tmean,
        "precipitation": list(properties.get("PRECTOTCORR", {}).values())[0] if properties.get("PRECTOTCORR") else np.nan,
        "wind_speed": list(properties.get("WS10M", {}).values())[0] if properties.get("WS10M") else np.nan,
        "wind_dir": list(properties.get("WD10M", {}).values())[0] if properties.get("WD10M") else np.nan,
        "solar_radiation": list(properties.get("ALLSKY_SFC_SW_DWN", {}).values())[0] if properties.get("ALLSKY_SFC_SW_DWN") else np.nan,
        "rh_mean": rh_mean,
    }

    # ----------------- Step 2: Derived Biological Indicators -----------------
    # (a) Vapour Pressure Deficit (VPD)
    if not np.isnan(tmean) and not np.isnan(rh_mean):
        es = 0.6108 * math.exp((17.27 * tmean) / (tmean + 237.3))  # Saturation vapor pressure (kPa)
        ea = es * rh_mean / 100  # Actual vapor pressure
        result["vpd"] = round(es - ea, 3)
    else:
        result["vpd"] = np.nan

    # (b) Growing Degree Days (GDD)
    if not np.isnan(tmin) and not np.isnan(tmax):
        gdd = max(((tmax + tmin) / 2) - base_temp, 0)
        result["gdd"] = round(gdd, 2)
    else:
        result["gdd"] = np.nan

    # ----------------- Step 3: Season Classification -----------------
    # Simple meteorological season (based on hemisphere)
    northern = lat >= 0
    if northern:
        if month in [12, 1, 2]:
            season = "Winter"
        elif month in [3, 4, 5]:
            season = "Spring"
        elif month in [6, 7, 8]:
            season = "Summer"
        else:
            season = "Autumn"
    else:
        if month in [12, 1, 2]:
            season = "Summer"
        elif month in [3, 4, 5]:
            season = "Autumn"
        elif month in [6, 7, 8]:
            season = "Winter"
        else:
            season = "Spring"

    result["season"] = season

    return result
```
