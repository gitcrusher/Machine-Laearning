# One-Hot Encoding with many categories in a feature

```python
import numpy as np 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df=pd.read_csv('mercedes.csv',usecols=["X0","X1","X2","X3","X4","X5","X6"])
df.head()
```

```python
for i in df.columns:
    print(len(df[i].unique()))
```

```python
df.X1.value_counts().sort_values(ascending = False).head(10)
```

```python
lst_10=df.X1.value_counts().sort_values(ascending = False).head(10).index
lst_10=list(lst_10)
```

```python
lst_10
```

```python
for categories in lst_10:
    df[categories]=np.where(df['X1']==categories,1,0)
```

```python
lst_10.append('X1')
```

```python
df[lst_10]
```

# count or frequency Encoding

```python
train_set = pd.read_csv('https://archive.ics.uci.edu/static/public/2/data.csv' , header = None,index_col=None)
train_set.head() 
```

```python
train_set.drop(0, inplace=True)
```

```python
train_set.head() 
```

```python
columns=[1,3,5,6,7,8,9,13]
train_set=train_set[columns]
train_set.columns=['Employment','degree','marital-status','designation','family_job','race','sex','country']
```

```python
train_set.head()
```

```python
for features in train_set.columns[:]:
    print(features,':',train_set[features].nunique(),'labels')
```

```python
country_map=train_set['country'].value_counts().to_dict()
```

#### here we can see that actually there is the case of same frequency 

```python
df_cols=pd.DataFrame(country_map.items(),columns=['country','count'])
```

```python
df_cols.head()
```

```python
df_cols.shape
```

```python
df_cols['country'].nunique()
```

```python
df_cols['count'].nunique()
```

```python
duplicate_rows=df_cols[df_cols['count'].duplicated(keep=False)]
duplicate_rows
```

```python
train_set['country']=train_set['country'].map(country_map)
train_set.head()
```

#### Advantages 
1) easy 
2) we are not increasing the feature space

#### Disadvantage
1) it will provide the same frequency count or weight if the frequencies are same 


# Target guided Ordinal Encoding
1. ordering the labels according to target
2. replace the labels by the joint probability of being 1 or 0

```python
df=pd.read_csv('titanic.csv',usecols=['Cabin','Survived'])
df.head()
```

```python
df['Cabin'].fillna('Missing',inplace=True)
```

```python
df['Cabin']=df['Cabin'].astype(str).str[0]
```

```python
df.head()
```

```python
df.Cabin.unique()
```

```python
df.groupby(['Cabin'])['Survived'].mean()
```

```python
df.groupby(['Cabin'])['Survived'].mean().sort_values().index
```

```python
ordinal_lables=df.groupby(['Cabin'])['Survived'].mean().sort_values().index
ordinal_lables
```

```python
enumerate(ordinal_lables,0)
```

```python
ordinal_lables2={k:i for i,k in enumerate(ordinal_lables,0)}
ordinal_lables2
```

```python
df['Cabin_ordinal_labels']=df['Cabin'].map(ordinal_lables2)
df.head()
```

# Mean Encoding

```python
mean_ordinal=df.groupby(['Cabin'])['Survived'].mean().to_dict()
```

```python
mean_ordinal
```

```python
df['mean_ordinal_encode']=df['Cabin'].map(mean_ordinal)
df.head()
```

#### Disadvantage
1. High Risk of Overfitting
2. Needs Smoothing
