```python
import ee
import pandas as pd
from tqdm import tqdm
import numpy as np

# Initialize Earth Engine
ee.Initialize(project='flowering-472417')

# ==========================
# Load dataset
# ==========================
df = pd.read_csv("final_dataset.csv")
df["date"] = pd.to_datetime(df[["Year", "Month", "Day"]])

# ==========================
# ERA5 dataset
# ==========================
era5 = ee.ImageCollection("ECMWF/ERA5/DAILY")

def get_weather_for_day(date, points):
    """Fetch climate for all points in one date (batch)"""
    start = ee.Date(date.strftime("%Y-%m-%d"))
    end = start.advance(1, "day")

    # Select ERA5 bands
    era5_day = era5.filterDate(start, end)
    if era5_day.size().getInfo() == 0:
        return pd.DataFrame()  # No data for this date

    img = era5_day.first().select([
        "minimum_2m_air_temperature",
        "maximum_2m_air_temperature",
        "dewpoint_temperature_2m",
        "total_precipitation"
    ])

    # Build FeatureCollection from points
    features = []
    for _, r in points.iterrows():
        point = ee.Geometry.Point([r["longitude"], r["latitude"]])
        features.append(ee.Feature(point, {
            "latitude": float(r["latitude"]),
            "longitude": float(r["longitude"]),
            "date": date.strftime("%Y-%m-%d")
        }))
    fc = ee.FeatureCollection(features)

    # Sample all points in one request
    sampled = img.sampleRegions(
        collection=fc,
        scale=10000,
        geometries=False
    ).getInfo()

    # Convert to pandas DataFrame
    records = []
    for f in sampled["features"]:
        props = f["properties"]
        records.append(props)
    return pd.DataFrame(records)

# ==========================
# Process in batches (by date)
# ==========================
all_weather = []
for date, group in tqdm(df.groupby("date"), desc="Fetching weather (batch)"):
    try:
        batch_df = get_weather_for_day(date, group)
        if not batch_df.empty:
            all_weather.append(batch_df)
    except Exception as e:
        print(f"⚠️ Skipping {date}: {e}")

# Combine all results
weather_df = pd.concat(all_weather, ignore_index=True)

# Merge back with original dataset
final_df = pd.merge(df, weather_df, on=["latitude", "longitude", "date"], how="left")

# Save locally
final_df.to_csv("dataset_with_weather.csv", index=False)
print("✅ Saved dataset_with_weather.csv")
```
