```python
import pandas as pd
import numpy as np
import math
import requests
from requests.adapters import HTTPAdapter, Retry
from concurrent.futures import ThreadPoolExecutor

# ----------------- SESSION WITH RETRIES -----------------
session = requests.Session()
retries = Retry(total=3, backoff_factor=1, status_forcelist=[500, 502, 503, 504])
session.mount("https://", HTTPAdapter(max_retries=retries))

# ----------------- HELPER FUNCTION -----------------
def get_first_value(d):
    if d:
        return next(iter(d.values()))
    return np.nan

# ----------------- NASA WEATHER FUNCTION (MONTHLY) -----------------
def fetch_weather_nasa_monthly(lat, lon, start_year=1984, end_year=2025, base_temp=9):
    # Skip invalid lat/lon
    if not (-90 <= lat <= 90) or not (-180 <= lon <= 180):
        print(f"⚠️ Invalid coordinates: {lat},{lon}")
        return None

    url = (
        f"https://power.larc.nasa.gov/api/temporal/monthly/point?"
        f"start={start_year}0101&end={end_year}1231"
        f"&latitude={lat}&longitude={lon}"
        "&parameters=T2M_MIN,T2M_MAX,T2M,PRECTOTCORR,WS10M,WD10M,ALLSKY_SFC_SW_DWN,RH2M"
        "&community=AG"
        "&format=JSON"
    )

    try:
        r = session.get(url, timeout=20)
        r.raise_for_status()
        data = r.json()
    except Exception as e:
        print(f"⚠️ Failed for {lat},{lon}: {e}")
        return None

    props = data.get("properties", {}).get("parameter", {})
    months = props.get("T2M", {}).keys()  # YYYYMM format
    weather_list = []

    for month in months:
        year = int(str(month)[:4])
        mon = int(str(month)[4:])
        tmin = get_first_value(props.get("T2M_MIN", {}).get(str(month), {}))
        tmax = get_first_value(props.get("T2M_MAX", {}).get(str(month), {}))
        tmean = get_first_value(props.get("T2M", {}).get(str(month), {}))
        rh_mean = get_first_value(props.get("RH2M", {}).get(str(month), {}))
        precipitation = get_first_value(props.get("PRECTOTCORR", {}).get(str(month), {}))
        wind_speed = get_first_value(props.get("WS10M", {}).get(str(month), {}))
        wind_dir = get_first_value(props.get("WD10M", {}).get(str(month), {}))
        solar_radiation = get_first_value(props.get("ALLSKY_SFC_SW_DWN", {}).get(str(month), {}))

        # VPD
        if not np.isnan(tmean) and not np.isnan(rh_mean):
            es = 0.6108 * math.exp((17.27 * tmean) / (tmean + 237.3))
            ea = es * rh_mean / 100
            vpd = round(es - ea, 3)
        else:
            vpd = np.nan

        # GDD
        if not np.isnan(tmin) and not np.isnan(tmax):
            gdd = max(((tmax + tmin)/2) - base_temp, 0)
            gdd = round(gdd,2)
        else:
            gdd = np.nan

        # Season
        northern = lat >= 0
        if northern:
            if mon in [12,1,2]: season = "Winter"
            elif mon in [3,4,5]: season = "Spring"
            elif mon in [6,7,8]: season = "Summer"
            else: season = "Autumn"
        else:
            if mon in [12,1,2]: season = "Summer"
            elif mon in [3,4,5]: season = "Autumn"
            elif mon in [6,7,8]: season = "Winter"
            else: season = "Spring"

        weather_list.append({
            "latitude": lat,
            "longitude": lon,
            "Year": year,
            "Month": mon,
            "tmin": tmin,
            "tmax": tmax,
            "tmean": tmean,
            "rh_mean": rh_mean,
            "precipitation": precipitation,
            "wind_speed": wind_speed,
            "wind_dir": wind_dir,
            "solar_radiation": solar_radiation,
            "vpd": vpd,
            "gdd": gdd,
            "season": season
        })
    return weather_list

# ----------------- FETCH WEATHER FOR ALL HOTSPOTS (PARALLEL) -----------------
def fetch_all(df_hotspots):
    all_data = []

    def worker(row):
        lat = row['latitude']
        lon = row['longitude']
        data = fetch_weather_nasa_monthly(lat, lon)
        if data:
            all_data.extend(data)

    with ThreadPoolExecutor(max_workers=5) as executor:
        executor.map(worker, [row for idx, row in df_hotspots.iterrows()])

    return pd.DataFrame(all_data)

# ----------------- CREATE HOTSPOTS DATAFRAME -----------------
# Replace 'df_final' with your main dataset containing 'latitude', 'longitude', 'Year', 'Month', 'Day'
df_hotspots = df_final[['latitude', 'longitude']].drop_duplicates().reset_index(drop=True)

# ----------------- FETCH WEATHER -----------------
df_weather = fetch_all(df_hotspots)

# ----------------- MERGE WITH ORIGINAL DATASET IF NEEDED -----------------
df_final = df_final.merge(df_weather, on=['latitude','longitude','Year','Month'], how='left')

print(df_weather.head())
print(df_final.head())
```
