```python
df_final['Day'] = df_final['observed_on'].apply(
    lambda x: x.split("/")[1] if "/" in x else x.split("-")[2]
).astype(int)

df_final['Month'] = df_final['observed_on'].apply(
    lambda x: x.split("/")[0] if "/" in x else x.split("-")[1]
).astype(int)

df_final['Year'] = df_final['observed_on'].apply(
    lambda x: x.split("/")[2] if "/" in x else x.split("-")[0]
).astype(int)
```
