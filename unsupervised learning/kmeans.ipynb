```python
import numpy as np 
```

```python
class KMeans:
    def __init__(self, k = 3, max_iters = 100):
        self.max_iters = max_iters
        self.centroids = None
        self.clusters = [[] ,for_ in range(self.k)]
    
    def fit(self,X):
        self.x= X
        self.n_samples, self.n_features = X.shape
        random_sample_inices = np.random.choice(self.n_samples, self.k, replace = False)
        for _ in range(self.max_iters):
            self.clusters = self._create_clusters(self.centroids)
            centriods.old = self.centroids
            self.centroids = self._get_centroids(self.clusters)
            if self._is_converged(centriods.old, self.centroids): 
                break

    def _create_clusters(self, centroids):
        clusters= [[] for _ in range(self.k)]

        for idx,sample in enumerate(self.x):
            centroid_idx = self._closest_centroid(sample,centroids)
            clusters[centroid_idx].append(idx)
        return clusters
    
    def _closest_centroid(self, sample, centroids):
        dist=[(np.sqrt(np.sum((sample - centroid) ** 2))) for centroid in centroids]
        return np.argmin(dist)

    def _get_centroids(self, clusters):
        centroids =np.zeros((self.k, self.n_features))
        for cluster_idx, cluster in enumerate(clusters):
            centroids[cluster_idx] = np.mean(self.x[list(cluster)], axis=0)
        return centroids
    

    def _is_converged(self, centroids_old, centroids_new):
        dist = [np.sqrt(np.sum((centroids_old[i] - centroids_new[i]) ** 2)) for i in  range(self.k)]
        return np.sum(dist) == 0
    
    def predict(self, X):
        predicted_labels = [self._closest_centroid(sample, self.centroids) for sample in X]
        return np.array(predicted_labels)
```
