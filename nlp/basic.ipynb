```python
sample_text = "is this a sample text for testing? <html >, <body>, <p> this is a paragraph </p> </body> </html>"
```

```python
import re 
def strip_words(data):
    p = re.compile(r'<.*?>')
    return p.sub('', data)
```

```python
strip_words(sample_text)
```
